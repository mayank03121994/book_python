### Problem 3: Is it a prime number?

Return True if a whole number is prime (greater than 1 and divisible only by 1 and itself).

```python
# Your attempt:
```

**Solution to problem 3**

```python
def is_prime(n):
    if n < 2:
        return False
    divisor = 2
    while divisor * divisor <= n:
        if n % divisor == 0:
            return False
        divisor += 1
    return True

print([n for n in range(1, 30) if is_prime(n)])
print(is_prime(2143))
```

### Problem 10: Palindrome check

Return True if text reads the same forwards and backwards, ignoring case, spaces and punctuation.

```python
# Your attempt:
```

**Solution to problem 10**

```python
def is_palindrome(text):
    cleaned = [char.lower() for char in text if char.isalnum()]
    left, right = 0, len(cleaned) - 1
    while left < right:
        if cleaned[left] != cleaned[right]:
            return False
        left += 1
        right -= 1
    return True

print(is_palindrome("Madam"))
print(is_palindrome("Never odd or even"))
print(is_palindrome("Priya"))
```

### Problem 17: String compression

Compress repeated characters: 'aaabbc' becomes 'a3b2c1'.

```python
# Your attempt:
```

**Solution to problem 17**

```python
def compress(text):
    if not text:
        return ""
    parts = []
    current = text[0]
    count = 1
    for char in text[1:]:
        if char == current:
            count += 1
        else:
            parts.append(f"{current}{count}")
            current = char
            count = 1
    parts.append(f"{current}{count}")
    return "".join(parts)

print(compress("aaabbc"))
print(compress("yyyyyn"))
print(compress(""))
```

### Problem 21: Merge two sorted lists

Merge two already-sorted lists into one sorted list, without calling sorted().

```python
# Your attempt:
```

**Solution to problem 21**

```python
def merge_sorted(first, second):
    i, j = 0, 0
    merged = []
    while i < len(first) and j < len(second):
        if first[i] <= second[j]:
            merged.append(first[i])
            i += 1
        else:
            merged.append(second[j])
            j += 1
    merged.extend(first[i:])
    merged.extend(second[j:])
    return merged

print(merge_sorted([2, 29, 1506], [1, 448, 2143, 102127]))
```

```python
def two_sum(numbers, target):
    seen = {}
    for position, number in enumerate(numbers):
        needed = target - number
        if needed in seen:
            return seen[needed], position
        seen[number] = position
    return None

print(two_sum([500, 1200, 300, 700], 1000))
print(two_sum([1, 2], 10))
```

### Problem 25: Best run of balance changes (maximum subarray)

**Solution to problem 25**

```python
def max_subarray_sum(changes):
    best = changes[0]
    current = changes[0]
    for change in changes[1:]:
        current = max(change, current + change)
        best = max(best, current)
    return best

daily_changes = [-200, 500, -100, 700, -1500, 300]
print(max_subarray_sum(daily_changes))
```

### Problem 34: Longest stretch without repeats (sliding window)

Return the length of the longest part of a string with no repeated characters.

```python
# Your attempt:
```

**Solution to problem 34**

```python
def longest_unique_stretch(text):
    last_seen = {}
    start = 0
    best = 0
    for position, char in enumerate(text):
        if char in last_seen and last_seen[char] >= start:
            start = last_seen[char] + 1
        last_seen[char] = position
        best = max(best, position - start + 1)
    return best

print(longest_unique_stretch("abcabcbb"))
print(longest_unique_stretch("management"))
print(longest_unique_stretch(""))
```

### Problem 35: Fast power (recursion)

Calculate x to the power n (n a whole number of 0 or more) with recursion, using as few multiplications as possible.

```python
# Your attempt:
```

**Solution to problem 35**

```python
def power(x, n):
    if n == 0:
        return 1
    half = power(x, n // 2)
    if n % 2 == 0:
        return half * half
    return half * half * x

print(power(2, 10))
print(round(power(1.07, 5), 4))
```

### Problem 36: Binary search

Find the position of a value in a sorted list, or return -1 if it is not there.

```python
# Your attempt:
```

**Solution to problem 36**

```python
def binary_search(sorted_items, target):
    low, high = 0, len(sorted_items) - 1
    while low <= high:
        middle = (low + high) // 2
        if sorted_items[middle] == target:
            return middle
        if sorted_items[middle] < target:
            low = middle + 1
        else:
            high = middle - 1
    return -1

balances = [-8019, 1, 2, 29, 448, 1506, 2143, 102127]
print(binary_search(balances, 1506))
print(binary_search(balances, 500))
```

Return every ordering of the characters in a short string.

```python
# Your attempt:
```

**Solution to problem 38**

```python
def permutations(text):
    if len(text) <= 1:
        return [text]
    result = []
    for i, char in enumerate(text):
        remaining = text[:i] + text[i + 1:]
        for ending in permutations(remaining):
            result.append(char + ending)
    return result

print(permutations("abc"))
```

### Problem 39: Valid brackets (stack)

```python
# Your attempt:
```

**Solution to problem 39**

```python
def brackets_balanced(text):
    pairs = {")": "(", "]": "[", "}": "{"}
    stack = []
    for char in text:
        if char in "([{":
            stack.append(char)
        elif char in pairs:
            if not stack or stack.pop() != pairs[char]:
                return False
    return not stack

print(brackets_balanced("{customer: [age, (job)]}"))
print(brackets_balanced("(]"))
print(brackets_balanced("(("))
```

### Problem 41: Design a bank account

Design a BankAccount class with deposit, withdraw (no overdraft) and a transaction history.

```python
# Your attempt:
```

**Solution to problem 41**

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self._balance = balance
        self.history = []

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self._balance += amount
        self.history.append(("deposit", amount))

    def withdraw(self, amount):
        if amount > self._balance:
            raise ValueError("Insufficient funds")
        self._balance -= amount
        self.history.append(("withdraw", amount))

account = BankAccount("Priya", 1506)
account.deposit(500)
account.withdraw(200)
print(account.balance)
print(account.history)
try:
    account.withdraw(10000)
except ValueError as error:
    print("Refused:", error)
```

### Problem 42: Implement a stack

Implement a Stack class with push, pop, peek, is_empty and len().

```python
# Your attempt:
```

**Solution to problem 42**

```python
class Stack:
    def __init__(self):
        self._items = []

    def push(self, item):
        self._items.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self._items.pop()

    def peek(self):
        if self.is_empty():
            return None
        return self._items[-1]

    def is_empty(self):
        return len(self._items) == 0

    def __len__(self):
        return len(self._items)

calls = Stack()
calls.push("Priya")
calls.push("Arjun")
print(calls.peek(), len(calls))
print(calls.pop(), calls.pop(), calls.is_empty())
```

### Problem 43: A call-centre queue

Model a first-come, first-served queue of customers waiting for a call back, using collections.deque.

```python
# Your attempt:
```

**Solution to problem 43**

```python
from collections import deque

class CallQueue:
    def __init__(self):
        self._waiting = deque()

    def add(self, customer):
        self._waiting.append(customer)

    def next_customer(self):
        if not self._waiting:
            return None
        return self._waiting.popleft()

    def __len__(self):
        return len(self._waiting)

queue = CallQueue()
for name in ["Priya", "Arjun", "Meera"]:
    queue.add(name)
print(queue.next_customer(), len(queue))
print(queue.next_customer(), queue.next_customer(), queue.next_customer())
```

### Problem 44: Loan products and EMI (polymorphism)

Design loan classes that each calculate their monthly instalment (EMI). A home loan uses the standard EMI formula; an interest-free loan just divides the amount.

```python
# Your attempt:
```

**Solution to problem 44**

```python
from abc import ABC, abstractmethod

class Loan(ABC):
    def __init__(self, amount, months):
        self.amount = amount
        self.months = months

    @abstractmethod
    def emi(self):
        pass

class InterestLoan(Loan):
    def __init__(self, amount, months, yearly_rate):
        super().__init__(amount, months)
        self.yearly_rate = yearly_rate

    def emi(self):
        r = self.yearly_rate / 12
        growth = (1 + r) ** self.months
        return round(self.amount * r * growth / (growth - 1), 2)

class InterestFreeLoan(Loan):
    def emi(self):
        return round(self.amount / self.months, 2)

loans = [InterestLoan(500000, 240, 0.085), InterestFreeLoan(60000, 12)]
for loan in loans:
    print(type(loan).__name__, loan.emi())
```

```python
def moving_average(values, k):
    window_sum = sum(values[:k])
    averages = [window_sum / k]
    for i in range(k, len(values)):
        window_sum += values[i] - values[i - k]
        averages.append(window_sum / k)
    return averages

first_balances = [c["balance"] for c in customers[:6]]
print(first_balances)
print([round(a, 1) for a in moving_average(first_balances, 3)])
```
