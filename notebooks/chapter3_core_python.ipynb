## 5. Mutable vs immutable
### Two names, one list

```python
balances = [100, 200]
backup = balances
backup.append(300)
print(balances)
```

### Making a real copy

```python
balances = [100, 200]
backup = balances.copy()
backup.append(300)
print(balances)
print(backup)
```

### One catch: `.copy()` is a shallow copy

```python
customers = [[58, 2143], [44, 29]]
backup = customers.copy()
backup[0][1] = 0
print(customers)
```

## 7. Loops
### The for loop

```python
balances = [2143, 29, 2, 1506, 1]

for balance in balances:
    print(balance)
```

### Building up a result inside a loop

```python
balances = [2143, 29, 2, 1506, 1]
total = 0

for balance in balances:
    total = total + balance

print("Total:", total)
print("Average:", total / len(balances))
```

### Combining loops and conditions: counting

```python
balances = [2143, 29, 2, 1506, 1]
low_count = 0

for balance in balances:
    if balance < 100:
        low_count += 1

print("Customers with balance under 100:", low_count)
```

### range()

```python
for attempt in range(1, 4):
    print("Calling customer, attempt", attempt)

print(list(range(5)))
print(list(range(1, 4)))
print(list(range(0, 10, 2)))
```

### The while loop

```python
attempts = 0
answered = False

while not answered:
    attempts += 1
    print("Call attempt", attempts)
    if attempts == 3:
        answered = True

print("Customer answered after", attempts, "calls")
```

### break

```python
balances = [2143, 29, -450, 1506, 1]

for balance in balances:
    if balance < 0:
        print("Found an overdrawn account:", balance)
        break
    print("Checked:", balance)
```

### continue

```python
balances = [2143, 29, -450, 1506, 1]

for balance in balances:
    if balance < 0:
        continue
    print("Positive balance:", balance)
```

## 8. Functions
### Your first function

```python
def describe_customer(name, age):
    message = f"{name} is {age} years old."
    return message

result = describe_customer("Priya", 33)
print(result)
print(describe_customer("Arjun", 47))
```

### Default values for parameters

```python
def add_interest(balance, rate=0.07):
    interest = balance * rate
    return round(balance + interest, 2)

print(add_interest(1000))
print(add_interest(1000, 0.05))
print(add_interest(rate=0.06, balance=2143))
```

### Returning several values

```python
def balance_summary(balances):
    lowest = min(balances)
    highest = max(balances)
    average = sum(balances) / len(balances)
    return lowest, highest, average

low, high, avg = balance_summary([2143, 29, 2, 1506, 1])
print(low, high, avg)
```

### Forgetting return: a classic bug

```python
def no_return(balance):
    balance + 500

print(no_return(1000))
```

### *args

```python
def total_deposits(*amounts):
    print("Amounts received:", amounts)
    return sum(amounts)

print(total_deposits(500, 1200, 300))
print(total_deposits(50))
```

### **kwargs

## 9. Lambda, map and filter
### Lambda

```python
def double(x):
    return x * 2

double_lambda = lambda x: x * 2

print(double(1506))
print(double_lambda(1506))
```

### map() and filter()

```python
balances = [2143, 29, 2, 1506, 1]

in_thousands = list(map(lambda b: b / 1000, balances))
print(in_thousands)

healthy = list(filter(lambda b: b > 1000, balances))
print(healthy)
```

### Sorting with a key

```python
customers = [
    {"name": "Priya", "balance": 1506},
    {"name": "Arjun", "balance": 29},
    {"name": "Meera", "balance": 2143},
]

by_balance = sorted(customers, key=lambda c: c["balance"], reverse=True)

for c in by_balance:
    print(c["name"], c["balance"])
```

## 10. Mini project: a loan-eligibility checker
### Step 1: Write the function

```python
def check_loan_eligibility(customer):
    if customer["age"] < 21 or customer["age"] > 60:
        return False, "Age outside 21 to 60"
    if customer["default"] == "yes":
        return False, "Has defaulted on credit before"
    if customer["loan"] == "yes":
        return False, "Already has a personal loan"
    if customer["balance"] < 500:
        return False, "Balance below 500"
    return True, "Eligible"
```

### Step 2: Test it on five customers

```python
sample_customers = [
    {"age": 58, "job": "management", "default": "no", "balance": 2143, "loan": "no"},
    {"age": 44, "job": "technician", "default": "no", "balance": 29, "loan": "no"},
    {"age": 33, "job": "entrepreneur", "default": "no", "balance": 2, "loan": "yes"},
    {"age": 47, "job": "blue-collar", "default": "no", "balance": 1506, "loan": "no"},
    {"age": 33, "job": "unknown", "default": "no", "balance": 1, "loan": "no"},
]

for number, customer in enumerate(sample_customers, start=1):
    eligible, reason = check_loan_eligibility(customer)
    print(f"Customer {number} ({customer['job']}): {reason}")
```

### Step 3: Run it on all 45,211 customers
This needs an internet connection to download the data. If the download fails, replace the first four lines with `import pandas as pd` and `bank = pd.read_csv("bank-full.csv", sep=";")` after uploading the file.

```python
!pip install ucimlrepo
from ucimlrepo import fetch_ucirepo

bank_data = fetch_ucirepo(id=222)
bank = bank_data.data.features.copy()

all_customers = bank.to_dict("records")
print("Customers loaded:", len(all_customers))

eligible_count = 0
reason_counts = {}

for customer in all_customers:
    eligible, reason = check_loan_eligibility(customer)
    if eligible:
        eligible_count += 1
    reason_counts[reason] = reason_counts.get(reason, 0) + 1

print(f"Eligible customers: {eligible_count:,} out of {len(all_customers):,}")
print(f"That is {eligible_count / len(all_customers):.1%}")
print()

for reason, count in sorted(reason_counts.items(), key=lambda item: item[1], reverse=True):
    print(f"{reason}: {count:,}")
```
