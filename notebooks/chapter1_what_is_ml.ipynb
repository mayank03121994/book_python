## The traditional way: hand-written rules

This is the rule-based spam filter from the section *Traditional programming vs machine learning*.

```python
def is_spam(email):
    if "lottery" in email.lower():
        return True
    if "click here to claim" in email.lower():
        return True
    if email.count("!") > 5:
        return True
    return False
```

## Test it on a few emails

The first two are obvious spam. The third is spam written to dodge the rules. The fourth is a genuine email from a very excited friend.

```python
emails = [
    "Congratulations! You have won the LOTTERY!",
    "Click here to claim your free phone",
    "You have won a l0ttery prize, cl!ck now",
    "We got the flat!!!!!! Party on Saturday!!",
]

for email in emails:
    print(is_spam(email), "->", email)
```

## What went wrong?

- The third email **is** spam, but the filter says `False`: changing one letter (`l0ttery`) was enough to fool it.
- The fourth email is **not** spam, but the filter says `True`, just because of the exclamation marks.

This is exactly why machine learning learns patterns from thousands of labelled examples instead of relying on hand-written rules. You will build a real classifier in Book 2.
