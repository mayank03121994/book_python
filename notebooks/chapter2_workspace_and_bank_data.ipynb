## Part C: Your first real dataset

### Cell 1: Install and import
(If you already ran the install cell above in this session, running it again is harmless.)

```python
!pip install ucimlrepo

from ucimlrepo import fetch_ucirepo
import pandas as pd
```

### Cell 2: Download the data
This needs an internet connection. If the download fails, see the fallback cell at the end of this notebook.

```python
bank_data = fetch_ucirepo(id=222)

X = bank_data.data.features
y = bank_data.data.targets
```

### Cell 3: How big is it?

```python
print("Features table:", X.shape)
print("Target table:", y.shape)
```

### Cell 4: Put it all in one table

```python
bank = X.copy()
bank["subscribed"] = y["y"]
```

### Cell 5: Look at the first few customers

```python
bank.head()
```

### Cell 6: Get an X-ray of the table with `.info()`
Remember: every column says it is complete, but many missing answers are hidden as the word "unknown".

```python
bank.info()
```

### Cell 7: Get the numbers at a glance with `.describe()`

```python
bank.describe()
```

### Cell 8: Ask your first questions

```python
print(bank["subscribed"].value_counts())
print()
print("Average age:", round(bank["age"].mean(), 1))
print("Youngest:", bank["age"].min(), " Oldest:", bank["age"].max())
```

### Bonus: counting the hidden "unknown" values
This is the check behind the *Watch out* box in the book.

```python
for column in ["job", "education", "contact", "poutcome"]:
    count = ((bank[column] == "unknown") | bank[column].isna()).sum()
    print(column, "->", count, "unknown or empty values")
```
