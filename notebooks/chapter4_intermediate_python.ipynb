## 2. Error handling
### try and except

````python
raw_ages = ["58", "44", "thirty-three", "47", ""]

for text in raw_ages:
    try:
        age = int(text)
        print("Age:", age)
    except ValueError:
        print("Could not read age:", repr(text))
````

### else and finally

````python
def average_balance(balances):
    try:
        result = sum(balances) / len(balances)
    except ZeroDivisionError:
        print("No balances given, cannot calculate an average.")
        return None
    except TypeError:
        print("Balances must all be numbers.")
        return None
    else:
        print("Average calculated successfully.")
        return round(result, 2)
    finally:
        print("Finished checking.")

print(average_balance([2143, 29, 1506]))
print()
print(average_balance([]))
print()
print(average_balance([2143, "29"]))
````

````python
class BankCustomer:
    def __init__(self, name, age, job, balance):
        self.name = name
        self.age = age
        self.job = job
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def describe(self):
        return f"{self.name}, {self.age}, {self.job}, balance {self.balance:,} euros"
````

## 5. The three big ideas of OOP
### Inheritance

````python
class BankCustomer:
    def __init__(self, name, age, balance):
        self.name = name
        self.age = age
        self.balance = balance

    def describe(self):
        return f"{self.name}: balance {self.balance:,} euros"

    def monthly_fee(self):
        return 5


class PremiumCustomer(BankCustomer):
    def __init__(self, name, age, balance, relationship_manager):
        super().__init__(name, age, balance)
        self.relationship_manager = relationship_manager

    def monthly_fee(self):
        return 0

    def describe(self):
        basic = super().describe()
        return f"{basic} (premium, manager: {self.relationship_manager})"


meera = PremiumCustomer("Meera", 47, 102127, "Mr. Rao")
priya = BankCustomer("Priya", 33, 1506)

print(meera.describe())
print(meera.age)
print(isinstance(meera, BankCustomer))
print(isinstance(priya, PremiumCustomer))
````

### Encapsulation

````python
class BankCustomer:
    def __init__(self, name, balance):
        self.name = name
        self._balance = balance

    def get_balance(self):
        return self._balance

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive")
        if amount > self._balance:
            raise ValueError(f"Insufficient funds: balance is {self._balance}")
        self._balance -= amount
        return self._balance

priya = BankCustomer("Priya", 1506)
print(priya.withdraw(500))

try:
    priya.withdraw(5000)
except ValueError as error:
    print("Error:", error)

print(priya.get_balance())
````

## 6. Why OOP matters in ML
### A model you build yourself

````python
class MostCommonAnswerModel:
    def __init__(self):
        self.answer_ = None

    def fit(self, answers):
        counts = {}
        for a in answers:
            counts[a] = counts.get(a, 0) + 1
        self.answer_ = max(counts, key=counts.get)
        return self

    def predict(self, how_many):
        return [self.answer_] * how_many

training_answers = ["no", "no", "yes", "no", "no", "yes", "no"]

model = MostCommonAnswerModel()
model.fit(training_answers)
print("Learned answer:", model.answer_)
print(model.predict(3))
````

````python
def add_interest(balance: float, rate_percent: float) -> float:
    """Return the balance after one year of simple interest.

    Args:
        balance: The starting balance in euros.
        rate_percent: The yearly interest rate, e.g. 7 for 7%.

    Returns:
        The new balance, rounded to 2 decimal places.
    """
    interest = balance * rate_percent / 100
    return round(balance + interest, 2)

print(add_interest(1000, 7))
````

### Type hints for collections

````python
def calculate_average(balances: list[float]) -> float:
    return sum(balances) / len(balances)

def count_jobs(jobs: list[str]) -> dict[str, int]:
    counts = {}
    for job in jobs:
        counts[job] = counts.get(job, 0) + 1
    return counts

print(calculate_average([2143, 29, 1506]))
print(count_jobs(["management", "technician", "management"]))
````

### Step 2: The class, the converter and the loader

````python
import csv
import json
from datetime import date


class BankCustomer:
    """One bank customer from the Bank Marketing dataset."""

    def __init__(self, age, job, balance, has_loan, subscribed):
        if age < 18 or age > 120:
            raise ValueError(f"Unrealistic age: {age}")
        self.age = age
        self.job = job
        self.balance = balance
        self.has_loan = has_loan
        self.subscribed = subscribed

    def is_overdrawn(self):
        """Return True if the balance is below zero."""
        return self.balance < 0

    def __str__(self):
        return f"BankCustomer(age={self.age}, job={self.job}, balance={self.balance})"


def customer_from_row(row):
    """Turn one CSV row (a dictionary of text) into a BankCustomer."""
    return BankCustomer(
        age=int(row["age"]),
        job=row["job"],
        balance=int(row["balance"]),
        has_loan=row["loan"] == "yes",
        subscribed=row["subscribed"] == "yes",
    )


def load_customers(path):
    """Read customers from a CSV file, skipping rows that cannot be read."""
    customers = []
    skipped = 0
    with open(path, newline="") as file:
        reader = csv.DictReader(file)
        for line_number, row in enumerate(reader, start=2):
            try:
                customers.append(customer_from_row(row))
            except ValueError as error:
                skipped += 1
                print(f"Skipped line {line_number}: {error}")
    return customers, skipped


customers, skipped = load_customers("bank_customers.csv")
print(f"Loaded {len(customers):,} customers, skipped {skipped}")
print(customers[0])
````

### Step 3: Analyse and save a report

````python
overdrawn = [c for c in customers if c.is_overdrawn()]
subscribers = [c for c in customers if c.subscribed]
average_balance = sum(c.balance for c in customers) / len(customers)

report = {
    "created_on": date.today().isoformat(),
    "customers_loaded": len(customers),
    "rows_skipped": skipped,
    "average_balance": round(average_balance, 2),
    "overdrawn_customers": len(overdrawn),
    "subscription_rate": round(len(subscribers) / len(customers), 3),
}

with open("customer_report.json", "w") as file:
    json.dump(report, file, indent=2)

with open("customer_report.json") as file:
    print(file.read())
````
